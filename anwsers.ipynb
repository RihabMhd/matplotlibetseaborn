```python
#challenge 2

import pandas as pd

import matplotlib.pyplot as plt

ventes_villes=pd.read_csv('./ventes_villes.csv')

plt.figure(figsize=(10, 5))
plt.bar(ventes_villes['ville'],ventes_villes['ventes'])

for i in range(len(ventes_villes)):
    plt.text(
        ventes_villes["ville"][i],
        ventes_villes["ventes"][i],
        ventes_villes["ventes"][i]
    )

plt.title("Ventes par ville")
plt.xlabel("Ville")
plt.ylabel("Ventes")

plt.grid(axis="y")
plt.show()
```

```python
#challenge 5

import pandas as pd
import matplotlib.pyplot as plt

entreprise=pd.read_csv('./entreprise.csv')

fig, ax = plt.subplots(2, 2) #fig=the whole figure

ax[0, 0].bar(
    entreprise["employe"],
    entreprise["salaire"]
)

ax[0, 0].set_title("Salaire par employé")
ax[0, 0].set_xlabel("Employé")
ax[0, 0].set_ylabel("Salaire")

ax[0, 1].hist(
    entreprise["salaire"],
    bins=5
)

ax[0, 1].set_title("Distribution des salaires")
ax[0, 1].set_xlabel("Salaire")
ax[0, 1].set_ylabel("Nombre d'employés")


ax[1, 0].scatter(
    entreprise["experience"],
    entreprise["performance"]
)

ax[1, 0].set_title("Expérience vs performance")
ax[1, 0].set_xlabel("Expérience")
ax[1, 0].set_ylabel("Performance")

moyenne = entreprise.groupby("departement")["salaire"].mean()

ax[1, 1].bar(
    moyenne.index,
    moyenne.values
)

ax[1, 1].set_title("Salaire moyen par département")
ax[1, 1].set_xlabel("Département")
ax[1, 1].set_ylabel("Salaire moyen")

plt.tight_layout()

plt.show()
```

```python
#challenge 6

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

clients=pd.read_csv('./clients.csv')

sns.histplot(data=clients, x="age", bins=5)

plt.title("distribution de l'âge")
plt.xlabel("Âge")
plt.ylabel("nombre de clients")
plt.show()

sns.countplot(data=clients, x="ville")

plt.title("nombre de clients par ville")
plt.xlabel("Ville")
plt.ylabel("Nombre de clients")
plt.show()

sns.boxplot(data=clients, x="salaire")

plt.title("distribution des salaires")
plt.xlabel("Salaire")
plt.show()

sns.barplot(data=clients, x="sexe", y="salaire")

plt.title("Salaire moyen selon le sexe")
plt.xlabel("Sexe")
plt.ylabel("Salaire moyen")
plt.show()
```

```python
#challenge 7

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

clients_depenses=pd.read_csv('./clients_depenses.csv')

sns.scatterplot(data=clients_depenses,x='salaire',y='depenses')
plt.title('relation entre depenses et salaire')
plt.xlabel('salaire')
plt.ylabel('depenses')
#plt.show()

sns.scatterplot(data=clients_depenses,x="nombre_commandes",y="depenses")

plt.title("relation entre nombre de commandes et dépenses")
plt.xlabel("Nombre de commandes")
plt.ylabel("Dépenses")
#plt.show()

sns.scatterplot(data=clients_depenses,x="age",y="salaire")

plt.title("relation entre âge et salaire")
plt.xlabel("Âge")
plt.ylabel("Salaire")
#plt.show()

sns.regplot(data=clients_depenses,x="salaire",y="depenses")

plt.title("Salaire et dépenses avec tendance")
plt.xlabel("Salaire")
plt.ylabel("Dépenses")
#plt.show()

sns.pairplot(
    clients_depenses[
        ["age", "salaire", "nombre_commandes", "depenses"]
    ]
)

plt.show()

#scatterplot() → see the relationship
#regplot() → see the relationship + trend
#pairplot() → explore many relationships at once
```
