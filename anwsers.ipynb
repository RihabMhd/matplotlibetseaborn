```python
#challenge 1

import pandas as pd

import matplotlib.pyplot as plt

# plt.plot()
# plt.title()
# plt.xlabel()
# plt.ylabel()
# plt.grid()
# plt.show()

temperature=pd.read_csv('./temperature.csv')

plt.plot(temperature['jour'],temperature['temperature'])
plt.title("temperature par jour")
plt.xlabel('jour')
plt.ylabel('temperature')

plt.grid()
plt.show()
```

```python
#challenge 2

import pandas as pd

import matplotlib.pyplot as plt

ventes_villes=pd.read_csv('./ventes_villes.csv')

plt.figure(figsize=(10, 5))
plt.bar(ventes_villes['ville'],ventes_villes['ventes'])

for i in range(len(ventes_villes)):
    plt.text(
        ventes_villes["ville"][i],
        ventes_villes["ventes"][i],
        ventes_villes["ventes"][i]
    )

plt.show()
```
